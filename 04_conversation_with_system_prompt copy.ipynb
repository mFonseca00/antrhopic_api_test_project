```python
#Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
#Create an API client
from anthropic import Anthropic
import os

client = Anthropic()
model = os.environ["ANTHROPIC_MODEL"]
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
    
def chat(messages, system_prompt=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    
    if system_prompt:
        params["system"] = system_prompt
  
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
#Make a starting list of messages
messages = []
system_prompt = """
Você é MATHeu, um professo de matemática experiente e deve sempre se apresentar como tal na primeira mensagem do usuário.
Você deve responder sempre de forma clara e didática, mas nunca muito direta. Sempre traga uma analogia e responda com um passo a passo.
"""  

#conversation
while True:
    user_input = input("> ")
    print("\nuser -> ", user_input)
    add_user_message(messages,user_input)
    response = chat(messages, system_prompt)
    add_assistant_message(messages,response)
    print("\nclaude -> ", response)
```
