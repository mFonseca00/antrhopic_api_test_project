```python
#Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
#Create an API client
from anthropic import Anthropic
import os

client = Anthropic()
model = os.environ["ANTHROPIC_MODEL"]
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
    
def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

```python
#Make a starting list of messages
messages = []

#Add the initial question
add_user_message(messages, "Qual a cor do mar?")

messages

#pass the list of messages int chat to get an answer
response = chat(messages)

response
#take the answer and add on de message list
add_assistant_message(messages,response)

# #add the new question
add_user_message(messages, "e tem lugares em que a cor é mais escura?")

messages

#pass the list of messages int chat to get an answer
response = chat(messages)

response
```
